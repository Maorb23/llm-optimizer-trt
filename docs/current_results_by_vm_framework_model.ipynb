# Current results: initial runs and parameters probe

November 17 collection only: `runs/20261117-143416/`, with executed notebooks under `experiments/notebooks/20261117-143416/` as evidence. Cached W&B snapshots and exports enrich **matching W&B IDs only**. No network access or new experiments.

The results follow **section → base model → framework → concurrency → objective**. Notebook version, phase, workload, artifact and revision, hardware, sampler, seed, and search space are recorded for each study. ModelOpt variants share the base model heading. Exact artifact IDs are available in rankings and run inventories.

Only runs with verified duration **≥ 300 seconds** contribute to every main result, ranking, timing, telemetry, plot, and importance calculation. Exactly 300 seconds is retained. Short, unknown-duration, conflicting, and mismatched records remain in the audit. Read-only SQLite connections and content/provenance deduplication retain genuine repeated evaluations.

Classification uses full W&B names: case-insensitive `invest` → **Parameters probe**, other known names → **Initial runs**. Missing or conflicting full names remain **unclassified**. All name evidence is retained; no collection name, notebook version, study name, or current `RUN_NAME` setting substitutes for full-name evidence. Within each cached W&B transaction log, the latest checksum-validated RunRecord name supersedes its initial placeholder; prior names remain in the evidence. Disagreements between independent sources remain unclassified, regardless of file iteration order.

The analysis cell computes the tables and both importance methods once. The report cells display the cached results.

Objective statistics use finite, feasible completed trials. Sample variance uses `ddof=1`, and singletons have unavailable variance. Each study contributes its best eligible completed trial to the ranking. Each ranking shows up to five studies. Output throughput is maximized and p95 E2E latency is minimized. All collected studies have scalar objectives.

Timing means use complete **execution attempts**, each distinct attempt ID contributing once per recorded phase. Only attempts joined by recorded attempt ID or explicit trial number enter the statistics; unassigned artifacts remain in the audit and no trial identity is inferred from row order. Copies of an attempt contribute once; retries with distinct IDs remain distinct. Per-trial startup curves first average complete attempts within a trial, then weight contributing trials equally. Objective curves keep original trial numbers and sampler/seed distinctions; failed and missing trials are never forward-filled. Count labels are observations/contributing runs. Engine-log observations may overlap and are never added to wall-clock phases. Launch-to-ready includes server initiation and model loading; missing telemetry is unavailable.

Server telemetry is measured over the recorded benchmark window. Coverage tables sum recorded sample counts and benchmark-window seconds within each objective and telemetry definition. Gauge means use covered-window weights, with availability flags and positive recorded coverage required; peaks stay separate. Analytic client requests in flight use their recorded window and have no scrape count/coverage fraction. Client concurrency is configuration, not server queue depth; GPU utilization is not kernel occupancy. Individual scrape samples are unavailable when the artifacts only persist window summaries. Detailed records and errors remain inside closed toggles.

The empirical v2/v3 schema comparison uses recorded framework-specific version metadata. The study metadata preserves recorded notebook labels and uncertainty.
